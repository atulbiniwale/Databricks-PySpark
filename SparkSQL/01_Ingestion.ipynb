
### Temp views?

- temp views are temporary. 
- They are deleted after the cluster is stopped/terminated/ after the session is over. 
- temp view is only available for that 1 session. 

now you can play around (temporarily) using the temp view created.

To make the temp view available to other sessions, create a Global Temp view.

```sql
%python
#creates a temp view "orders_temp"

df.createOrReplaceTempView("sales_temp")
```

```sql
select * from sales_temp limit 4
```

#### Global temp View

- temp views which are available in all sessions till the cluster is on/working.
- they are only available in all purpose compute and other computes (but not serverless compute)
