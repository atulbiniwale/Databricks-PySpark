### Ranking

```python
%sql
select Item_Type, Item_MRP as price_per_unit,
        Rank() over ( partition by Item_Type order by Item_MRP desc) as rank,
        Dense_Rank() over ( partition by Item_Type order by Item_MRP desc) as dense_rank,
        Row_Number() over ( partition by Item_Type order by Item_MRP desc) as row_number
from sparksql_catelog.sparksql_schema.sales_managedtable
where Item_MRP < 38
order by Item_MRP desc
```

### Framing / Partitions for Cumulative totals

```python
%sql
select  Item_Type, Item_MRP, 
        sum(Item_MRP) over (order by Item_MRP rows between unbounded preceding and current row) as RunningTotal1,
        sum(Item_MRP) over (order by Item_MRP rows between unbounded preceding and unbounded following) as Total1,
        -- combinations can be done between unbounded and preceding/following, current row
        
        sum(Item_MRP) over (partition  by Item_Type order by Item_MRP rows between unbounded preceding and current row) as RunningTotal2,
        sum(Item_MRP) over (partition by Item_Type order by Item_MRP rows between unbounded preceding and unbounded following) as Total2
       
from sparksql_catelog.sparksql_schema.sales_managedtable
```
