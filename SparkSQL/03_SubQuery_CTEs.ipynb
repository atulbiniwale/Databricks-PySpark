### SubQuery

```python
%sql
select * from sparksql_catelog.sparksql_schema.sales_managedtable limit 5

--Subquery (inside Where, from mainly or in joins )

--1. inside query
select * from sparksql_catelog.sparksql_schema.sales_managedtable where order_status = 'Delivered' and product_category = 'Fashion'

--2. outside query
select * 
from (
  select * from sparksql_catelog.sparksql_schema.orders_managedtable where order_status = 'Delivered' and product_category = 'Fashion'
) as tbl1
where payment_method = 'PayPal'
```

### CTEs

```python
WITH tbl2 AS (

WITH tbl1 AS (
  SELECT * FROM sparksql_catelog.sparksql_schema.orders_managedtable WHERE order_status = 'Delivered' AND product_category = 'Fashion'
)
SELECT * FROM tbl1 WHERE payment_method = 'PayPal' -- outer query using CTE.

) 
SELECT * FROM tbl2 where product_name = 'Sunglasses'
```
