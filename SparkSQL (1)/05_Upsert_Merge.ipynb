### Upsert or Merge (means Update plus Insert)

this is Very Very Imp

```sql
%python
df = spark.read.table("sparksql_catelog.sparksql_schema.sales_managedtable")
```

```sql
%python

# CREATE A TEMP VIEW
df.createOrReplaceTempView("sales_temp" )

# or (use any of the methods for creating a temp view)

# CREATE A TEMP VIEW (alternative way to create a temp view)
spark.sql('''SELECT * FROM {sales_temp}''', sales_temp = df)
```

### if you have new sales records, then you can Upsert them in other table.

```sql
MERGE INTO sparksql_catelog.sparksql_schema.sales_managedtable  as target
USING sales_temp as source
ON target.Item_Identifier = source.Item_Identifier
WHEN MATCHED THEN
UPDATE SET *
WHEN NOT MATCHED
THEN INSERT *
```
